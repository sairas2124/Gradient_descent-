```python
import requests

url = "https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page=1"

response = requests.get(url)
data = response.json()

# Extract movie overviews (text data)
texts = [movie['overview'] for movie in data['results']]

# Take one sample
text = texts[0]
print(text)
```

```python
import re
import nltk

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Chat word dictionary
chat_words = {
    "u": "you",
    "r": "are",
    "lol": "laughing",
    "omg": "oh my god"
}

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def clean_text(text):

    # 1. Lowercase
    text = text.lower()

    # 2. Remove HTML tags
    text = re.sub(r'<.*?>', '', text)

    # 3. Remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)

    # 4. Remove punctuation and special characters
    text = re.sub(r'[^a-z\s]', '', text)

    # 5. Chat word treatment
    words = text.split()
    words = [chat_words.get(word, word) for word in words]

    # 6. Tokenization
    tokens = word_tokenize(" ".join(words))

    # 7. Remove stopwords
    tokens = [word for word in tokens if word not in stop_words]

    # 8. Lemmatization
    tokens = [lemmatizer.lemmatize(word) for word in tokens]

    return " ".join(tokens)
```

```python
cleaned_texts = [clean_text(t) for t in texts]

# Show result
print("Original:\n", texts[0])
print("\nCleaned:\n", cleaned_texts[0])
```

METHODS OF CONVETRING TEXT INTO NUMBER

```python
#BoW(bag your words)
from sklearn.feature_extraction.text import CountVectorizer

texts = ["I love NLP", "I love AI"]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)

print(X.toarray())
```

```python
#Bigram
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(ngram_range=(2,2))  # only bigrams
X = vectorizer.fit_transform(texts)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
#TD-IDF
from sklearn.feature_extraction.text import TfidfVectorizer

texts = ["I love NLP", "I love AI"]

vectorizer = TfidfVectorizer()
#vectorizer = TfidfVectorizer(ngram_range=(1,2))
X = vectorizer.fit_transform(texts)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```
